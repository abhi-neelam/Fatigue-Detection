```python
num_participants = 18
```

```python
for pid in range(1, num_participants+1):
    if pid == 4 or pid == 16:
        continue

    for protocol in ["control", "fatigue"]:
        interim_file_names = glob.glob(f"data/DUO-GAIT/interim/OG_st_{protocol}/sub_{pid:02}/*.csv")

        starting_sample_indices = []
        for interim_file in interim_file_names:
            df = pd.read_csv(interim_file)
            df.rename({ "timestamp": "Time (secs)", "Unnamed: 0": "Sample" }, axis=1, inplace=True)
            starting_sample_indices.append(int(df.iloc[0]['Sample']))
        starting_sample_indices = np.array(starting_sample_indices)
        if not np.all(starting_sample_indices == starting_sample_indices[0]):
            print(pid, starting_sample_indices)
```
