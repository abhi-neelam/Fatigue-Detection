```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings
import os
import numpy as np
warnings.filterwarnings('ignore')
```

```python
downsample_factor = 100 # to reduce plotly plot memory sizes
```

Visualize only for 1st participant

```python
hr_df_01 = pd.read_csv("data/DUO-GAIT/raw/OG_st_raw/sub_01/heart_rate.CSV",skiprows=2)
hr_df_01 = hr_df_01.filter(["Time", "HR (bpm)"], axis=1)
hr_df_01["Time"] = pd.to_datetime(hr_df_01["Time"])
hr_df_01.rename({"Time":"Time (min)"},axis=1,inplace=True)

hr_df_07 = pd.read_csv("data/DUO-GAIT/raw/OG_st_raw/sub_07/heart_rate.CSV",skiprows=2)
hr_df_07 = hr_df_07.filter(["Time", "HR (bpm)"], axis=1)
hr_df_07["Time"] = pd.to_datetime(hr_df_07["Time"])
hr_df_07.rename({"Time":"Time (min)"},axis=1,inplace=True)

fig = px.line(hr_df_07.iloc[::downsample_factor], x='Time (min)', y="HR (bpm)", title="Fit Male HR Plot")
fig.show()

fig = px.line(hr_df_07.iloc[::downsample_factor], x='Time (min)', y="HR (bpm)", title="Fit Female HR Plot")
fig.show()
```

From here, it's clear that the middle portion is when the fatiguing protocol is in effect for the male. At the edges, the participants walk at their self selected pace. Even at the last time, the HR never dips below 90 indicating that HR can be used to classify fatigue for this particular participant. Given the resting heart rate is close to 60 for this person and on the lower end of lactate while fatigued, it's likely this male is a fit person. Similar situation goes for the female participant. Do other participants exhibit this similar behavior? Is there any participant that doesn't exhibit this behavior?

```python
fig = px.line(lw_df_01.iloc[::downsample_factor], x='Time (secs)', y=["AccX", "AccY", "AccZ"], title="Fit Male Acc Plot for Left Wrist")
fig.show()

fig = px.line(rw_df_01.iloc[::downsample_factor], x='Time (secs)', y=["AccX", "AccY", "AccZ"], title="Fit Male Acc Plot for Right Wrist")
fig.show()

fig = px.line(lw_df_01.iloc[::downsample_factor], x='Time (secs)', y=["GyrX", "GyrY", "GyrZ"], title="Fit Male Gyr Plot for Left Wrist")
fig.show()

fig = px.line(rw_df_01.iloc[::downsample_factor], x='Time (secs)', y=["GyrX", "GyrY", "GyrZ"], title="Fit Male Gyr Plot for Right Wrist")
fig.show()
```

Firstly, there are periods of ACC which are constant. This indicates that no movements occur here. These are clues for transition periods to next tasks such as the fatigue protocol and eventually the fatiguing walk. Another indication of this is big movements which are shown as spikes in the ACC plots. Additionally, the heart rate and IMU graphs are clearly not synchronized so it's hard to merge them without timestamps. On top of this, the slight variants of ACC spikes for opposite wrists indicate asymetrical data which is expected (even for Y axis). The data is also extremely noisy so have to be careful.
