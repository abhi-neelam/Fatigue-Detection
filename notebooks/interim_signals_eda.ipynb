```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings
import os
import numpy as np
warnings.filterwarnings('ignore')
```

```python
pd.set_option('display.max_columns', 200)
pd.set_option('display.width', 150)
```

```python
downsample_factor = 50 # to reduce plotly plot memory sizes
```

```python
while not os.path.isdir(os.path.join(os.getcwd(), 'data')):
    os.chdir("../") # set cwd to root dir
```

```python
lw_control_df_02 = pd.read_csv("data/DUO-GAIT/interim/OG_st_control/sub_02/LW.csv")
lw_control_df_02.rename({ "timestamp": "Time (secs)", "Unnamed: 0": "Sample" }, axis=1, inplace=True)
lw_control_df_02['AccM'] = np.linalg.norm(lw_control_df_02[['AccX', 'AccY', 'AccZ']].values, axis=1)
lw_control_df_02['GyrM'] = np.linalg.norm(lw_control_df_02[["GyrX", "GyrY", "GyrZ"]].values, axis=1)

lw_fatigue_df_02 = pd.read_csv("data/DUO-GAIT/interim/OG_st_fatigue/sub_02/LW.csv")
lw_fatigue_df_02.rename({ "timestamp": "Time (secs)", "Unnamed: 0": "Sample" }, axis=1, inplace=True)
lw_fatigue_df_02['AccM'] = np.linalg.norm(lw_fatigue_df_02[['AccX', 'AccY', 'AccZ']].values, axis=1)
lw_fatigue_df_02['GyrM'] = np.linalg.norm(lw_fatigue_df_02[["GyrX", "GyrY", "GyrZ"]].values, axis=1)

ll_control_df_02 = pd.read_csv("data/DUO-GAIT/interim/OG_st_control/sub_02/LL.csv")
ll_control_df_02.rename({ "timestamp": "Time (secs)", "Unnamed: 0": "Sample" }, axis=1, inplace=True)
ll_control_df_02['AccM'] = np.linalg.norm(ll_control_df_02[['AccX', 'AccY', 'AccZ']].values, axis=1)
ll_control_df_02['GyrM'] = np.linalg.norm(ll_control_df_02[["GyrX", "GyrY", "GyrZ"]].values, axis=1)

ll_fatigue_df_02 = pd.read_csv("data/DUO-GAIT/interim/OG_st_fatigue/sub_02/LL.csv")
ll_fatigue_df_02.rename({ "timestamp": "Time (secs)", "Unnamed: 0": "Sample" }, axis=1, inplace=True)
ll_fatigue_df_02['AccM'] = np.linalg.norm(ll_fatigue_df_02[['AccX', 'AccY', 'AccZ']].values, axis=1)
ll_fatigue_df_02['GyrM'] = np.linalg.norm(ll_fatigue_df_02[["GyrX", "GyrY", "GyrZ"]].values, axis=1)
```

```python
fig = px.line(lw_control_df_02.iloc[::downsample_factor], x='Time (secs)', y=["AccX", "AccY", "AccZ"], title="Avg Male Acc Plot for Control (Left Wrist)")
fig.show()

fig = px.line(lw_fatigue_df_02.iloc[::downsample_factor], x='Time (secs)', y=["AccX", "AccY", "AccZ"], title="Avg Male Acc Plot for Fatigue (Left Wrist)")
fig.show()

fig = px.line(lw_control_df_02.iloc[::downsample_factor], x='Time (secs)', y=["GyrX", "GyrY", "GyrZ"], title="Avg Male Gyr Plot for Control (Left Wrist)")
fig.show()

fig = px.line(lw_fatigue_df_02.iloc[::downsample_factor], x='Time (secs)', y=["GyrX", "GyrY", "GyrZ"], title="Avg Male Gyr Plot for Fatigue (Left Wrist)")
fig.show()
```

```python
fig = px.line(ll_control_df_02.iloc[::downsample_factor], x='Time (secs)', y=["AccX", "AccY", "AccZ"], title="Avg Male Acc Plot for Control (Left Leg)")
fig.show()

fig = px.line(ll_fatigue_df_02.iloc[::downsample_factor], x='Time (secs)', y=["AccX", "AccY", "AccZ"], title="Avg Male Acc Plot for Fatigue (Left Leg)")
fig.show()

fig = px.line(ll_control_df_02.iloc[::downsample_factor], x='Time (secs)', y=["GyrX", "GyrY", "GyrZ"], title="Avg Male Gyr Plot for Control (Left Leg)")
fig.show()

fig = px.line(ll_fatigue_df_02.iloc[::downsample_factor], x='Time (secs)', y=["GyrX", "GyrY", "GyrZ"], title="Avg Male Gyr Plot for Fatigue (Left Leg)")
fig.show()
```

Differences are faint between control and fatigue. For wrist gyroZ + accY movements, it seems the magnitudes are damped for fatigue state compared to control. For left leg, accX is also damped. 

```python
fig = px.line(lw_control_df_02.iloc[::downsample_factor], x='Time (secs)', y=["AccM"], title="Avg Male Acc Magnitude for Control (Left Wrist)")
fig.show()

fig = px.line(lw_fatigue_df_02.iloc[::downsample_factor], x='Time (secs)', y=["AccM"], title="Avg Male Acc Magnitude for Fatigue (Left Wrist)")
fig.show()

fig = px.line(lw_control_df_02.iloc[::downsample_factor], x='Time (secs)', y=["GyrM"], title="Avg Male Gyr Magnitude for Control (Left Wrist)")
fig.show()

fig = px.line(lw_fatigue_df_02.iloc[::downsample_factor], x='Time (secs)', y=["GyrM"], title="Avg Male Gyr Magnitude for Fatigue (Left Wrist)")
fig.show()
```

```python
fig = px.line(ll_control_df_02.iloc[::downsample_factor], x='Time (secs)', y=["AccM"], title="Avg Male Acc Magnitude for Control (Left Leg)")
fig.show()

fig = px.line(ll_fatigue_df_02.iloc[::downsample_factor], x='Time (secs)', y=["AccM"], title="Avg Male Acc Magnitude for Fatigue (Left Leg)")
fig.show()

fig = px.line(ll_control_df_02.iloc[::downsample_factor], x='Time (secs)', y=["GyrM"], title="Avg Male Gyr Magnitude for Control (Left Leg)")
fig.show()

fig = px.line(ll_fatigue_df_02.iloc[::downsample_factor], x='Time (secs)', y=["GyrM"], title="Avg Male Gyr Magnitude for Fatigue (Left Leg)")
fig.show()
```

Acc Magnitude for left leg shows many more high spikes for fatigue compared to control. This isn't the case for gyro as it looks very similar across the tasks. For wrist, acc magnitude also shows more spikes. For gyro, magnitudes are higher for control compared to fatigue.
